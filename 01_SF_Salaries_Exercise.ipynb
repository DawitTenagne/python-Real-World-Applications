** Import pandas as pd.**

```python
import pandas as pd
```

** Read Salaries.csv as a dataframe called sal.**

```python
from numpy import nan
sal = pd.read_csv("Salaries.csv")
sal
```

```python
sal.head(10)
```

```python
sal.info()
```

**What is the average BasePay ?**

```python
sal["BasePay"] = pd.to_numeric(sal["BasePay"], errors="coerce")
sal["BasePay"] = sal["BasePay"].fillna(sal["BasePay"].mean())
averSal = sal["BasePay"].mean()
print(averSal)
```

** What is the highest amount of OvertimePay in the dataset ? **

```python
sal["OvertimePay"] = pd.to_numeric(sal["OvertimePay"], errors = "coerce")
sal["OvertimePay"].max()
```

** What is the job title of  JOSEPH DRISCOLL ? Note: Use all caps, otherwise you may get an answer that doesn't match up (there is also a lowercase Joseph Driscoll). **

```python
joseph_df = sal[sal["EmployeeName"] == "JOSEPH DRISCOLL"] # data frame of that specific employer only
joseph_df["JobTitle"]
```

** How much does JOSEPH DRISCOLL make (including benefits)? **

```python
joseph_df["TotalPayBenefits"]
```

** What is the name of highest paid person (including benefits)?**

```python
sal["TotalPayBenefits"] = pd.to_numeric(sal["TotalPayBenefits"], errors = "coerce")
#highest_pay = sal["TotalPayBenefits"].max()
# sal[sal["TotalPayBenefits"] == highest_pay]["EmployeeName"]

# there is also a shortcut and more andvanced way to do it.
sal.iloc[sal["TotalPayBenefits"].argmax()] # or sal.iloc[sal["TotalPayBenefits"].idxmax()]

# argmax() returns the index (row label) of the maximum value in a Series.
# argmin() returns the index (row label) of the smallest value in a Series.

```

** What is the name of lowest paid person (including benefits)? Do you notice something strange about how much he or she is paid?**

```python
sal["TotalPayBenefits"] = pd.to_numeric(sal["TotalPayBenefits"], errors = "coerce") # to change into numbers and to convert any string into NAN
# sal.iloc[sal['TotalPayBenefits'].argmin()] # The advanced way
min_pay = sal['TotalPayBenefits'].min()
sal[sal['TotalPayBenefits'] == min_pay]
```

** What was the average (mean) BasePay of all employees per year? (2011-2014) ? **

```python
sal["BasePay"] = pd.to_numeric(sal["BasePay"], errors= "coerce")
sal.groupby("Year")["BasePay"].mean()
```

** How many unique job titles are there? **

```python
sal["JobTitle"].unique() # return an array of unique values in the given column
sal["JobTitle"].nunique() # return the number of unique values in the given column
```

** What are the top 5 most common jobs? **

```python
sal["JobTitle"].value_counts().head(5)
```

** How many Job Titles were represented by only one person in 2013? (e.g. Job Titles with only one occurence in 2013?) **

```python
jobTitleCount_df = (sal[sal['Year']==2013])["JobTitle"].value_counts() # it return a series (a single column)
sum(jobTitleCount_df == 1)
# jobTitleCount_df == 1 --> return a boolean series
#in python, sum(True, false, True) = return the number of true value
# a series is like a python series with more features
```

** How many people have the word Chief in their job title? (This is pretty tricky) **

```python
# we need to make a custom function

def chiefChecker(title):
  if "chief" in title.lower().split():
    return True
  else:
    return False

chiefboolean = sal["JobTitle"].apply(lambda x:chiefChecker(x)) # return boolean series

sum(chiefboolean)
```

** Bonus: Is there a correlation between length of the Job Title string and Salary? **

```python
# adding new column named title_len = that tell use the length of the title

sal["title_len"] = sal["JobTitle"].apply(len)  # it apply the length function to each string under the title column


# df.corr() = to get the correlation
corr = sal[['TotalPayBenefits',"title_len"]].corr()
corr
```

```python
import seaborn as sns
sns.heatmap(corr, cmap="coolwarm")
```
